## 🧩 Step 3: Basic Data Cleaning

### 3.1 Correct Sentinel Values and Impossible Dates

**🎯 Objective:**
Apply cleaning rules to replace sentinel values and correct impossible dates.

**Instructions:**
- In `age`, replace the **-999** sentinel with the median.
- In `city`, replace the `"?"` sentinel with missing values (`pd.NA`).
- Mark out-of-range dates as missing (`pd.NA`).


```python
# Replace -999 with the median age
age_mediana = users['age'].median()
users['age'] = users['age'].replace(-999, age_mediana)

# Verify changes
users['age'].describe()
```

```python
# Replace ? with NA in city
users['city'] = users['city'].replace('?', np.nan)

# Verify changes
print((users['city'] == '?').sum())
```

```python
# Dates from 2026 before replacement
print("\n2026 dates before replacement")
print((users['reg_date'].dt.year == 2026).sum())

# Mark future reg_date values as NA
users.loc[users['reg_date'].dt.year == 2026, 'reg_date'] = pd.NaT

# Verify changes
print("\n2026 dates after replacement")
print((users['reg_date'].dt.year == 2026).sum())
print("\nMissing dates after replacement")
print(users['reg_date'].isna().sum())
```

## 🧩 Step 4: User-Level Usage Summary Statistics

### 4.1 Aggregate Usage Behavior

🎯 **Objective:** Summarize the key variables in the `usage` table **by user**, creating metrics that represent each customer's actual historical usage behavior.

**Instructions:**
1. Build an aggregated `usage` table by `user_id` that includes:
   - total number of messages
   - total number of calls
   - total call minutes

2. Rename the columns using clear technical names:
   - `cant_mensajes`
   - `cant_llamadas`
   - `cant_minutos_llamada`

3. Merge this table with `users`.


```python
# Helper columns
usage["is_text"] = (usage["type"] == "text").astype(int) # total number of messages
usage["is_call"] = (usage["type"] == "call").astype(int) # total number of calls


# Aggregate information by user
usage_agg = usage.groupby("user_id").agg({
    "is_text": "sum",
    "is_call": "sum",
    "duration": "sum"
}).reset_index()

# Review the result
usage_agg.head(3)
```

```python
# Rename columns
usage_agg = usage_agg.rename(columns={
    "is_text": "cant_mensajes",
    "is_call": "cant_llamadas",
    "duration": "cant_minutos_llamada"
})

# Review the result
usage_agg.head(3)
```

```python
# Merge the aggregated table with the users dataset

user_profile = users.merge(
    usage_agg,
    on="user_id",
    how="left"
)

user_profile.head(5)
```

### 5.2 Outlier Identification

🎯 **Objective:**
Detect extreme values in key **usage** and **customer** variables that could affect the analysis and decide whether they require cleaning or further review.

**Instructions:**
- Use **boxplots** to visually identify outliers in the following columns:
  - `age`
  - `cant_mensajes`
  - `cant_llamadas`
  - `cant_minutos_llamada`
- Create a **for loop** to generate the four boxplots automatically.

- After creating the charts, state whether **outliers are present** in each variable.
- If there are outliers, create another loop to calculate the limits for those columns using the **IQR method** and decide how to handle them.
  - If outliers occur on only one side, it is not necessary to calculate both limits.

**Hint:**
- Inside the loop, use `plt.title(f'Boxplot: {col}')` so the title changes according to the column.


```python
# Visualize using boxplots
columnas_numericas = ['age', 'cant_mensajes', 'cant_llamadas', 'cant_minutos_llamada']

for col in columnas_numericas:
    plt.figure(figsize=(8,4))
    plt.boxplot(user_profile[col].dropna())
    plt.title(f'Boxplot: {col}')
    plt.ylabel(col)
    plt.grid(True, alpha=0.3)
    plt.show()
```

💡 **Insights:**

- `age`: No outliers are present. All ages appear to fall within a reasonable range for the analyzed sample.

- `cant_mensajes`: Outliers are present only at the upper end of the distribution. This suggests the presence of a few users who are much more active in sending messages than the rest of the sample.

- `cant_llamadas`: Outliers are also present only at the upper end of the distribution. This suggests the existence of a small group of users with substantially more intensive call usage than the typical behavior in the sample.

- `cant_minutos_llamada`: Significant high-end outliers are present. However, these values likely represent users with intensive service usage rather than data errors. Removing them is not recommended because they are part of genuine customer behavior and could be important for analyzing plan usage.


```python
# Calculate limits using the IQR method
columnas_limites = ['age','cant_mensajes','cant_llamadas','cant_minutos_llamada']

for col in columnas_limites:

    Q1 = user_profile[col].quantile(0.25)
    Q3 = user_profile[col].quantile(0.75)

    IQR = Q3 - Q1

    limite_inferior = Q1 - 1.5 * IQR
    limite_superior = Q3 + 1.5 * IQR

    outliers = user_profile[
        (user_profile[col] < limite_inferior) |
        (user_profile[col] > limite_superior)
    ]

    print(f'\n{col}')
    print(f'Lower bound: {limite_inferior:.2f}')
    print(f'Upper bound: {limite_superior:.2f}')
    print(f'Number of outliers: {len(outliers)}')
```

```python
# Review the upper limits and maximum values to decide whether to keep the outliers
user_profile[columnas_limites].describe()
```

💡 **Insights:**

`cant_mensajes`: Keep the outliers?

✅ Keep them.

The maximum values reach 17 messages, while 75% of users send 7 messages or fewer. Although some users are more active, these values are plausible and represent real usage behavior rather than data-entry errors.

`cant_llamadas`: Keep the outliers?

✅ Keep them.

The maximum is 15 calls, compared with a median of 4 calls. Users with many calls are a minority, but they still represent legitimate service-usage patterns. Removing them could bias the analysis of the most active customers.

`cant_minutos_llamada`: Keep the outliers?

✅ Keep them, but analyze them carefully.

Although very high values exist (up to 155.69 minutes) compared with a median of 19.78 minutes, these records may correspond to users with intensive service usage. There is no evidence of errors or impossible values, so keeping them is preferable in order to preserve relevant information about the highest-usage customers.

## 🧩 Step 6: Customer Segmentation

### 6.1 Customer Segmentation by Usage

🎯 **Objective:** Classify each user into a usage group (Low Usage, Medium Usage, High Usage) based on the number of calls and messages recorded.

**Instructions:**
- Create a new column named `grupo_uso` in the `user_profile` DataFrame.
- Use logical comparisons (`<`, `>`) to evaluate call and message conditions and assign:
  - `'Low Usage'` when calls < 5 and messages < 5
  - `'Medium Usage'` when calls < 10 and messages < 10
  - `'High Usage'` for all remaining cases.


```python
# Create the grupo_uso column
condiciones = [
    (user_profile['cant_llamadas'] < 5) & (user_profile['cant_mensajes'] < 5),
    (user_profile['cant_llamadas'] < 10) & (user_profile['cant_mensajes'] < 10)
]

valores = [
    'Low Usage',
    'Medium Usage'
]

user_profile['grupo_uso'] = np.select(
    condiciones,
    valores,
    default='High Usage'
)

```

```python
# Verify changes
user_profile.head()
```

### 6.2 Customer Segmentation by Age

🎯 **Objective:** Classify each user into an **age group**.

**Instructions:**
- Create a new column named `grupo_edad` in the `user_profile` DataFrame.
- Use logical comparisons (`<`, `>`) to evaluate the conditions and assign:
  - `'Young'` when `age < 30`
  - `'Adult'` when `age < 60`
  - `'Senior'` for all remaining cases.


```python
# Create the grupo_edad column
condiciones = [
    user_profile['age'] < 30,
    user_profile['age'] < 60
]

valores = [
    'Young',
    'Adult'
]

user_profile['grupo_edad'] = np.select(
    condiciones,
    valores,
    default='Senior'
)

```
